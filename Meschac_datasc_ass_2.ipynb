```python
fig, ax = plt.subplots(figsize=(10, 6))
ax.scatter(world['GDP per capita (constant 2010 US$)'], world['Mortality rate, under-5 (per 1,000 live births)'], alpha=0.5)
ax.set_xlabel('GDP per capita (constant 2010 US$)')
ax.set_ylabel('Mortality rate, under-5 (per 1,000 live births)')
ax.set_title('Under-5 Mortality Rate vs. GDP per Capita')
plt.show()
```

```python
# Create log-transformed variables
world['log_gdp_per_cap'] = np.log(world['GDP per capita (constant 2010 US$)'])
world['log_under5_mortality_rate'] = np.log(world['Mortality rate, under-5 (per 1,000 live births)'])
```

```python
world.head()
```

```python
world.columns
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
ax.scatter(world['log_gdp_per_cap'], world['log_under5_mortality_rate'], alpha=0.5, color="green")
ax.set_xlabel('Log of GDP per capita (constant 2010 US$)')
ax.set_ylabel('Log of Mortality rate, under-5 (per 1,000 live births)')
ax.set_title('Log Under-5 Mortality Rate vs. Log GDP per Capita')
ax.legend()
plt.show()
```

```python
# Calculate and display the correlation coefficient
correlation = world['log_gdp_per_cap'].corr(world['log_under5_mortality_rate'])
print(f"The Pearson correlation coefficient between log GDP per capita and log under-5 mortality rate is: {correlation}")
```

The graph shows a very strong negative correlation between the Log of GDP per Capita and the Log of the Under-5 Mortality Rate. The Pearson correlation coefficient of $-0.839$ is very close to $-1$, indicating that as GDP per capita increases, the child mortality rate decreases, in a very significant and linear fashion. In essence, there is strong evidence that a country's wealth is closely linked to the survival of its children. The data points are tightly clustered and clearly follow this downward trend.

```python
# Fit a linear regression model
from sklearn.linear_model import LinearRegression

# Drop rows with NaN values in the relevant columns for the regression
world_cleaned = world.dropna(subset=['log_gdp_per_cap', 'log_under5_mortality_rate'])

X = world_cleaned['log_gdp_per_cap'].values.reshape(-1, 1)
y = world_cleaned['log_under5_mortality_rate'].values.reshape(-1, 1)

model = LinearRegression()
model.fit(X, y)

# Predict the y values for the line
y_pred = model.predict(X)

# Create the scatter plot with the regression line
fig, ax = plt.subplots(figsize=(10, 6))
ax.scatter(world_cleaned['log_gdp_per_cap'], world_cleaned['log_under5_mortality_rate'], alpha=0.5, color="green", label='Data points')
ax.plot(X, y_pred, color='red', linewidth=2, label='Regression line')

ax.set_xlabel('Log of GDP per capita (constant 2010 US$)')
ax.set_ylabel('Log of Mortality rate, under-5 (per 1,000 live births)')
ax.set_title('Log Under-5 Mortality Rate vs. Log GDP per Capita with Regression Line')
ax.legend()

plt.show()
```

To "pass a line through the points" and visualize the correlation, I used Linear Regression (the Ordinary Least Squares method OLS).This approach was the most appropriate because the scatter plot exhibits a clearly linear trend and the correlation coefficient of $-0.839$ is very strong. The regression line I plotted represents the best possible fit to the data by minimizing the sum of the squared errors, which allows us to accurately model and quantify this strong negative correlation between the Log of GDP and the Log of the mortality rate.

# For this exercice i will use HSI.The ticker HSI on Yahoo Finance represents the Hang Seng Index, which is the benchmark index for the Hong Kong Stock Exchange. It tracks the daily movements of the largest companies listed on the Hong Kong Stock Exchange and serves as the main indicator of overall market performance.

```python
# Define the stock index ticker and date range
ticker = '^HSI'
start_date = '2010-01-01'
end_date = '2025-10-16'

# Download the data
HK = yf.download(ticker, start=start_date, end=end_date)

# Save the data to a CSV file
file_name = f"{ticker[1:]}.csv"  # Remove the '^' for the filename
HK.to_csv(file_name)

print(f"Downloaded data for {ticker} and saved to {file_name}")
```

```python
HK['Returns'] = HK['Close'].pct_change()
display(HK.head())
```

```python
# Change subplot layout to a 2x2 grid
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(15, 15))
plt.suptitle("Stock Data Analysis (HK)", fontsize=20, fontweight='bold')

# --- Plot 1: Close Price -> axes[0, 0] ---
# Ensure column names are correct for your DataFrame HK
axes[0, 0].plot(HK.index, HK[('Close', '^HSI')], color='blue')
axes[0, 0].set_title('Closing Price', fontsize=15)
axes[0, 0].set_ylabel('Close Price')
axes[0, 0].grid(True)

# --- Plot 2: Volume -> axes[0, 1] ---
# Ensure column names are correct for your DataFrame HK
axes[0, 1].plot(HK.index, HK[('Volume', '^HSI')], color='Orange')
axes[0, 1].set_title('Trading Volume', fontsize=15)
axes[0, 1].set_ylabel('Volume')
axes[0, 1].grid(True)

# --- Plot 3: Daily Returns (Line Plot) -> axes[1, 0] ---
# Reduced linewidth to better show the often small daily variations
axes[1, 0].plot(HK.index, HK['Returns'], color='Green', linewidth=0.5) # thinner line
axes[1, 0].set_title('Daily Returns (Time Series)', fontsize=15)
axes[1, 0].set_ylabel('Daily Returns')
axes[1, 0].set_xlabel('Date')
axes[1, 0].grid(True)

# --- Plot 4: Histogram of Returns -> axes[1, 1] ---
axes[1, 1].hist(HK['Returns'].dropna(), bins=50, color='Purple', alpha=0.8)
axes[1, 1].set_title('Distribution of Daily Returns', fontsize=15)
axes[1, 1].set_ylabel('Frequency')
axes[1, 1].set_xlabel('Daily Returns')
axes[1, 1].grid(axis='y', alpha=0.6)

# Final adjustment for better layout, leaving space for suptitle
plt.tight_layout(rect=[0, 0, 1, 0.97])
```
