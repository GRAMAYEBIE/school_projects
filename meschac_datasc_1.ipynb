```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import yfinance as yf
```

```python
TICKER = "TSLA" # Tesla
START_DATE = "2020-01-01"
END_DATE = "2025-10-08"
```

```python
TSLA = yf.download(TICKER, start=START_DATE, end=END_DATE)
print(f" We downloded {TICKER} and place the data in 'TSLA'.")
```
